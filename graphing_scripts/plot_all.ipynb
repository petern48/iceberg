```python
%matplotlib inline
import json
import os

from utils import (
    DEFAULT_DATA,
    DEFAULT_OUT,
    plot_pruning_read_row_groups,
    plot_pruning_read_datafiles,
    plot_disk_storage,
    plot_memory_read,
    plot_memory_write,
    plot_time_read,
    plot_time_write,
)
```

```python
display_inline = True  # Set to False to save to disk instead
input_dir = DEFAULT_DATA
output_dir = DEFAULT_OUT


with open(input_dir) as f:
    data = json.load(f)

os.makedirs(output_dir, exist_ok=True)
```

```python
if display_inline:
    print("Displaying graphs inline")
else:
    print(f"Generating graphs → {output_dir}/")


plot_pruning_read_row_groups(data, output_dir, display_inline=display_inline)
plot_pruning_read_datafiles(data, output_dir, display_inline=display_inline)
plot_disk_storage(data, output_dir, display_inline=display_inline)
plot_memory_read(data, output_dir, display_inline=display_inline)
plot_memory_write(data, output_dir, display_inline=display_inline)
plot_time_read(data, output_dir, display_inline=display_inline)
plot_time_write(data, output_dir, display_inline=display_inline)
print("Done.")
```
